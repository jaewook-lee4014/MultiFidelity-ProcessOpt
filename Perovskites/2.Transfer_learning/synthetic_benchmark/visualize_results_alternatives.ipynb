```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 실제 데이터 로드
df = pd.read_csv('benchmark_no_blr_20260115_055151/results.csv')
print(df.head())
print(f"Benchmarks: {df['benchmark'].unique()}")
print(f"Models: {df['model'].unique()}")
```

```python
def plot_strip(df_bench, benchmark_name, ax):
    model_order = df_bench.groupby('model')['final_regret'].mean().sort_values().index.tolist()
    
    sns.stripplot(data=df_bench, y='model', x='final_regret', order=model_order,
                  color='steelblue', alpha=0.7, size=10, jitter=0.15, ax=ax)
    
    means_df = df_bench.groupby('model')['final_regret'].mean().reindex(model_order)
    for i, (model, mean) in enumerate(means_df.items()):
        ax.scatter(mean, i, color='red', s=200, marker='D', zorder=5, edgecolor='black', linewidth=1.5)
        ax.text(mean + 0.03, i, f'{mean:.4f}', va='center', fontsize=9, fontweight='bold')
    
    ax.set_xlabel('Final Regret', fontsize=12)
    ax.set_ylabel('')
    ax.set_title(f'{benchmark_name} ({len(df_bench["seed"].unique())} seeds)\n◆ = Mean', fontsize=13, fontweight='bold')
    ax.grid(axis='x', alpha=0.3)
```

## Strip Plot: Branin-Fav & Branin-Unfav

```python
fig, axes = plt.subplots(1, 2, figsize=(20, 8))

# Branin-Fav
df_fav = df[df['benchmark'] == 'Branin-Fav']
plot_strip(df_fav, 'Branin-Fav', axes[0])

# Branin-Unfav
df_unfav = df[df['benchmark'] == 'Branin-Unfav']
plot_strip(df_unfav, 'Branin-Unfav', axes[1])

plt.tight_layout()
plt.savefig('stripplot_both.png', dpi=150)
plt.show()
```

```python
def plot_violin(df_bench, benchmark_name, ax):
    model_order = df_bench.groupby('model')['final_regret'].mean().sort_values().index.tolist()
    
    sns.violinplot(data=df_bench, y='model', x='final_regret', order=model_order,
                   palette='Blues', inner=None, ax=ax)
    sns.stripplot(data=df_bench, y='model', x='final_regret', order=model_order,
                  color='black', alpha=0.5, size=5, ax=ax)
    
    means_df = df_bench.groupby('model')['final_regret'].mean().reindex(model_order)
    for i, (model, mean) in enumerate(means_df.items()):
        ax.scatter(mean, i, color='red', s=150, marker='o', zorder=5, edgecolor='white', linewidth=2)
        ax.text(mean + 0.03, i, f'{mean:.4f}', va='center', fontsize=9, fontweight='bold')
    
    ax.set_xlabel('Final Regret', fontsize=12)
    ax.set_ylabel('')
    ax.set_title(f'{benchmark_name} ({len(df_bench["seed"].unique())} seeds)\n● = Mean', fontsize=13, fontweight='bold')
    ax.grid(axis='x', alpha=0.3)
```

## Violin Plot: Branin-Fav & Branin-Unfav

```python
fig, axes = plt.subplots(1, 2, figsize=(20, 8))

df_fav = df[df['benchmark'] == 'Branin-Fav']
plot_violin(df_fav, 'Branin-Fav', axes[0])

df_unfav = df[df['benchmark'] == 'Branin-Unfav']
plot_violin(df_unfav, 'Branin-Unfav', axes[1])

plt.tight_layout()
plt.savefig('violin_both.png', dpi=150)
plt.show()
```

```python
def plot_box(df_bench, benchmark_name, ax):
    model_order = df_bench.groupby('model')['final_regret'].mean().sort_values().index.tolist()
    
    sns.boxplot(data=df_bench, y='model', x='final_regret', order=model_order,
                color='lightgray', width=0.5, ax=ax)
    sns.stripplot(data=df_bench, y='model', x='final_regret', order=model_order,
                  color='steelblue', alpha=0.7, size=8, ax=ax)
    
    means_df = df_bench.groupby('model')['final_regret'].mean().reindex(model_order)
    for i, (model, mean) in enumerate(means_df.items()):
        ax.scatter(mean, i, color='red', s=150, marker='X', zorder=5, edgecolor='black', linewidth=0.5)
        ax.text(mean + 0.03, i, f'{mean:.4f}', va='center', fontsize=9, fontweight='bold')
    
    ax.set_xlabel('Final Regret', fontsize=12)
    ax.set_ylabel('')
    ax.set_title(f'{benchmark_name} ({len(df_bench["seed"].unique())} seeds)\nX = Mean', fontsize=13, fontweight='bold')
    ax.grid(axis='x', alpha=0.3)
```

## Box Plot: Branin-Fav & Branin-Unfav

```python
fig, axes = plt.subplots(1, 2, figsize=(20, 8))

df_fav = df[df['benchmark'] == 'Branin-Fav']
plot_box(df_fav, 'Branin-Fav', axes[0])

df_unfav = df[df['benchmark'] == 'Branin-Unfav']
plot_box(df_unfav, 'Branin-Unfav', axes[1])

plt.tight_layout()
plt.savefig('boxplot_both.png', dpi=150)
plt.show()
```
